## 1. Pandasをはじめよう
`pd.read_csv` でデータを読み込みます。


```python
import pandas as pd

# CSVの読み込み
# エラーが出る場合は、左側のファイルメニューに 'sales_data.csv' があるか確認してください
df = pd.read_csv("sales_data.csv")

print(df.head())
print(df.describe())
```

## 2. データの抽出 (Filtering)
条件式を使ってデータを絞り込みます。


```python
# 1000円以上の商品
high_price = df[df['price'] >= 1000]
print(high_price)

# 複数条件
target = df[(df['price'] >= 1000) & (df['stock'] > 0)]
```

## 3. データの加工と集計
### 列の計算


```python
df['total_value'] = df['price'] * df['stock']
print(df.head())
```

### グループ化集計 (Group By)


```python
summary = df.groupby('Category')['Sales'].sum()
print(summary)
```
