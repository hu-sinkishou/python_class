## 条件分岐とループで「アルゴリズム」を構築する


## 1. 比較演算子と真偽値
* `==`: 等しい
* `!=`: 等しくない
* `>`: より大きい
* `<`: より小さい
* `>=`: 以上
* `<=`: 以下


```python
x = 10
print(x == 10)
print(x > 20)
```

## 2. 論理演算子
* `and`: かつ
* `or`: または
* `not`: 〜ではない


```python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("入場できます")
else:
    print("入場できません")
```

## 3. 条件分岐 (If Statement)
インデント（字下げ）が重要です。


```python
score = 85

if score >= 90:
    print("評価：S")
elif score >= 70:
    print("評価：B")
elif score >= 60:
    print("評価：C")
else:
    print("評価：D")
```

## 4. 繰り返し (For Loop)
決まった回数のループ処理に使います。


```python
# 3回繰り返す
for i in range(3):
    print(f"{i}回目のハロー！")
```

### 合計値を計算する


```python
total = 0
for i in range(1, 101):
    total = total + i
print(f"1から100までの合計: {total}")
```

## 5. 条件による繰り返し (While Loop)
条件が満たされている間はずっと繰り返します。無限ループに注意してください。


```python
hp = 3
while hp > 0:
    print(f"勇者は戦っている (HP: {hp})")
    hp = hp - 1
print("勇者は倒れた...")
```

## 6. ループの制御 (Break & Continue)
* `break`: 中断して脱出
* `continue`: スキップして次へ


```python
for i in range(1, 11):
    if i == 5:
        print("5を見つけました！終了します。")
        break
    print(i)
```
