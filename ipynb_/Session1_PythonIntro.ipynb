## 0. はじめに：なぜPythonなのか？

**Pythonの3つの強み**

## 1. 画面への表示 (Print)
プログラミングの第一歩は、画面に文字を表示させることです。`print()` 関数を使います。


```python
# 文字列 (String) - クォーテーションで囲む
print("Hello, Python!")
print('こんにちは、世界')

# 数値 (Number) - そのまま書く
print(100)
print(3.14)

# 計算式 - 計算結果が表示される
print(10 + 20)
print(100 * 1.1)
```

**注意点**：`"1" + "1"` は文字の結合（11）になり、`1 + 1` は数値の計算（2）になります。


## 2. 変数 (Variables)
データを入れる「名前付きの箱」です。

**変数のルール**
1. `変数名 = 値` で代入する（右から左へ）。
2. 変数名は英小文字とアンダースコアが一般的。
3. 数字から始まる名前はNG。


```python
# 代入
apple_price = 100
count = 3

# 計算
total_price = apple_price * count

# 表示
print(total_price)
```

### 変数の上書き
中身を入れ替えることができます。


```python
hp = 100
print(hp)

hp = 50    # ダメージを受けた
print(hp)

hp = hp + 20 # 回復した
print(hp)
```

## 3. データ型 (Data Types)
* **int**: 整数 (10, -5)
* **float**: 小数 (3.14, -2.5)
* **str**: 文字列 ("Hello")


```python
print(type(10))
print(type(3.14))
print(type("Hello"))
```

## 4. 演算とf-strings

### 算術演算子
* `+`, `-`, `*`, `/`: 四則演算
* `//`: 商（整数部分）
* `%`: 余り
* `**`: べき乗


```python
print(10 / 3)
print(10 // 3)
print(10 % 3)
print(2 ** 3)
```

### f-strings（フォーマット文字列）
文字列の前に `f` をつけ、変数を `{ }` で囲みます。


```python
name = "田中"
age = 19
height = 175.5

print(f"私は {name} です。年齢は {age} 歳です。")
print(f"身長は {height} cmです。")
```

## 5. ユーザー入力 (Input)
**重要**: `input()` で受け取ったデータは全て「文字列」になります。計算に使う場合は `int()` や `float()` で変換が必要です。


```python
# 文字列として受け取る
name = input("あなたの名前は？: ")
print(f"ようこそ、{name}さん！")

# 数値として受け取り計算する
age_str = input("年齢を入力してください: ")
age = int(age_str)
print(f"来年は {age + 1} 歳ですね。")
```
