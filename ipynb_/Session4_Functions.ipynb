## 1. 関数 (Function) の基本
DRY原則（Don't Repeat Yourself）のために、処理をまとめます。


```python
def tasizan(a, b):
    answer = a + b
    return answer

result = tasizan(3, 5)
print(result)
```

### PrintとReturnの違い
* `print`: 表示するだけ。再利用できない。
* `return`: 値を返す。変数に入れたり計算に使える。


## 2. 引数の高度な使い方
デフォルト引数とキーワード引数。


```python
def calc_price(price, tax=1.1):
    return int(price * tax)

print(calc_price(1000))
print(calc_price(1000, 1.08))
```

## 3. 複数の値を返す


```python
def get_stats(scores):
    return max(scores), min(scores), sum(scores) / len(scores)

data = [10, 20, 30, 40, 50]
h, l, a = get_stats(data)
print(f"最高:{h}, 最低:{l}, 平均:{a}")
```
