# 第3回：データをまとめて扱う
## リストと辞書で大量のデータを管理する


## 1. リスト (List)
データを一列に並べて管理します。インデックスは0から始まります。


```python
members = ["佐藤", "鈴木", "高橋", "田中"]
print(members[0])
print(members[-1])
```

### 追加・削除・スライス


```python
members.append("渡辺")
members.remove("鈴木")
print(members[0:3])
```

### 便利な関数


```python
nums = [10, 20, 30, 40, 50]
print(max(nums))
print(min(nums))
print(sum(nums))
```

## 2. リストとFor文
リストの中身を順に取り出して処理します。


```python
prices = [100, 200, 150, 300]
total = 0
for p in prices:
    print(f"現在の商品: {p}円")
    total += p
print(f"合計: {total}円")
```

## 3. 辞書 (Dictionary)
キーと値のペアで管理します。


```python
student = {
    "name": "山田太郎",
    "age": 19,
    "univ": "A大学"
}
print(student["name"])
```

### 辞書の操作とループ


```python
student["weight"] = 65

menu = {"カレー": 500, "ラーメン": 600, "うどん": 400}
for name, price in menu.items():
    print(f"{name} は {price} 円です")
```
